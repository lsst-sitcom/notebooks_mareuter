```python
import pathlib

import matplotlib.pyplot as plot
import numpy as np
import rawpy
```

```python
data_path = pathlib.Path("/scratch/users/m/mreuter/LFA")
```

```python
files = sorted(list(data_path.glob("*.cr2")))
```

```python
indicies = []
medians = []
for ifile in files:
    index = int(ifile.stem[-4:])
    indicies.append(index)
    with rawpy.imread(str(ifile)) as raw:
        rgb = raw.postprocess(no_auto_bright=True, use_auto_wb=False, gamma=(1, 1), output_bps=16)
        luminance = np.dot(rgb[..., :3], [0.299, 0.587, 0.114])
        median = np.median(luminance)
        medians.append(median)
```

```python
plot.plot(np.array(indicies), np.array(medians))
```
